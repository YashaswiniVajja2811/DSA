```python
import heapq

def dijkstra(graph, start, num_nodes):
    heap = [(0, start)]
    distances = {i: float('inf') for i in range(num_nodes)}
    distances[start] = 0

    while heap:
        current_distance, node = heapq.heappop(heap)
        for neighbor, weight in graph[node]:
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(heap, (distance, neighbor))

    return distances
```

```python
def get_user_graph():
    num_nodes = int(input("Enter the number of nodes: "))
    graph = {i: [] for i in range(num_nodes)}

    num_edges = int(input("Enter the number of edges: "))
    print("Enter the edges in the format: source destination weight")

    for _ in range(num_edges):
        src, dest, weight = map(int, input().split())

        # Check if src and dest are within the valid range of nodes
        if 0 <= src < num_nodes and 0 <= dest < num_nodes:
            graph[src].append((dest, weight))
            graph[dest].append((src, weight))  # For undirected graph
        else:
            print(f"Invalid edge: Node {src} or {dest} is out of range (0-{num_nodes - 1}). Skipping this edge.")

    return graph, num_nodes
```

```python
6# Get user input
graph, num_nodes = get_user_graph()
start_node = int(input("Enter the starting node: "))

# Run Dijkstra's Algorithm
shortest_distances = dijkstra(graph, start_node, num_nodes)
```

```python
# Print shortest distances from start node
print("\nShortest distances from node", start_node)
for node, distance in shortest_distances.items():
    print(f"Node {node}: {distance}")
```
