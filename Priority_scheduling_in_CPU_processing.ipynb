```python
import heapq

class MaxHeap:
    def __init__(self):
        self.heap = []

    def push(self, val):
        heapq.heappush(self.heap, -val)  # Negate value for max heap

    def pop(self):
        return -heapq.heappop(self.heap)  # Negate back

    def top(self):
        return -self.heap[0] if self.heap else None

# Taking user input
heap = MaxHeap()
nums = list(map(int, input("Enter numbers to insert into max heap: ").split()))
for num in nums:
    heap.push(num)

print("Max Element:", heap.top())
heap.pop()
print("After removing max, new max:", heap.top())
```
